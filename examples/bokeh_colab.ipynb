```python
import sys
import livelossplot
print('livelossplot:', livelossplot.__version__)
print('running in Colab:', 'google.colab' in sys.modules)
```

```python
from time import sleep
import numpy as np

from livelossplot import PlotLosses
from livelossplot.outputs import BokehPlot

plotlosses = PlotLosses(outputs=[BokehPlot()])

for i in range(10):
    plotlosses.update({
        'acc': 1 - np.random.rand() / (i + 2.),
        'val_acc': 1 - np.random.rand() / (i + 0.5),
        'loss': 1. / (i + 2.),
        'val_loss': 1. / (i + 0.5),
    })
    plotlosses.send()
    sleep(0.5)
```

## What you should see

Two Bokeh panels (Accuracy, Loss) that **redraw with growing data lines** every ~0.5s. Before this fix, the panels rendered as empty skeletons and never updated (issue #109).
